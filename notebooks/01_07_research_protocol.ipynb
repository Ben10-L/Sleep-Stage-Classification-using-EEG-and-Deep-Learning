## 2. Verify group separation on explicitly synthetic subject IDs

```python
import numpy as np
from eeg_sleep.splits import subject_split
groups=np.repeat([f"subject_{i:02d}" for i in range(12)],5)
tr,va,te=subject_split(groups)
print({"train_subjects":len(set(groups[tr])),"val_subjects":len(set(groups[va])),"test_subjects":len(set(groups[te]))})
assert set(groups[tr]).isdisjoint(groups[te])
assert set(groups[va]).isdisjoint(groups[te])
```
