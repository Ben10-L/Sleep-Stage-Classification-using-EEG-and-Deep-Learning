## 3. Literature quality checklist

- Is subject-level leakage excluded?
- Are original source annotation conventions stated?
- Are class imbalance and N1 performance reported?
- Is the code or dataset selection reproducible?
- Do the authors explain whether EOG/EMG are used?
- Are confidence intervals or multiple random seeds reported?
