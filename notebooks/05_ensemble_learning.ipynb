## 1. Uniform soft voting

For `M` models and class `k`, averaged probability is `p_ens(k|x) = (1/M) Σ_m p_m(k|x)`. Weights are fixed before inspecting test results.

```python
import numpy as np
p1=np.array([[.7,.1,.1,.05,.05],[.1,.6,.1,.1,.1]])
p2=np.array([[.6,.1,.2,.05,.05],[.2,.5,.1,.1,.1]])
p_ensemble=(p1+p2)/2
print("Ensemble probabilities:", p_ensemble)
print("Predicted class IDs:",p_ensemble.argmax(axis=1))
assert np.allclose(p_ensemble.sum(axis=1),1)
```
