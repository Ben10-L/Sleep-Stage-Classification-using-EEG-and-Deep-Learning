## 1. Sleep stage labels

Wake is a classification label but **not** a sleep stage. N1, N2 and N3 belong to NREM sleep. REM is associated with characteristic eye movements and mixed-frequency EEG.

```python
from eeg_sleep.constants import STAGES, STAGE_TO_INT, ANNOTATION_TO_STAGE
for label in STAGES:
    print(f"{STAGE_TO_INT[label]}: {label}")
```

## 2. Label mapping is a research choice

The source hypnogram uses historical R&K stages. Grouping source stages 3 and 4 as N3 is a conversion for the five-class machine-learning task, **not** a new AASM expert annotation.

```python
for source, target in ANNOTATION_TO_STAGE.items():
    print(f"{source:16s} → {target}")
```

## 3. Draw a simple *illustrative* hypnogram

This sequence is manually invented for teaching. It is not a physiological recording.

```python
import numpy as np
import matplotlib.pyplot as plt
sequence = ["W","N1","N2","N3","N2","REM","W","N1","N2","REM"]
values = [STAGE_TO_INT[x] for x in sequence]
fig, ax = plt.subplots(figsize=(10, 3))
ax.step(np.arange(len(values)+1)*30, values+[values[-1]], where="post")
ax.set_yticks(range(5), STAGES)
ax.set_xlabel("Elapsed time (seconds)")
ax.set_ylabel("Stage label")
ax.set_title("Illustrative hypnogram (not real subject data)")
ax.invert_yaxis(); plt.show()
```
